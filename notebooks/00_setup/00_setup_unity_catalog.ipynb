```python
%sql
-- 00_setup_unity_catalog : architecture Unity Catalog du projet LivrEx
-- Idempotent : peut être relancé sans erreur (IF NOT EXISTS)

CREATE CATALOG IF NOT EXISTS livraison
  COMMENT 'Projet LivrEx : plateforme analytics livraison de repas';

USE CATALOG livraison;

CREATE SCHEMA IF NOT EXISTS landing COMMENT 'Zone de dépôt des fichiers sources';
CREATE SCHEMA IF NOT EXISTS bronze  COMMENT 'Données brutes historisées en Delta';
CREATE SCHEMA IF NOT EXISTS silver  COMMENT 'Données nettoyées, typées, dédoublonnées';
CREATE SCHEMA IF NOT EXISTS gold    COMMENT 'Tables analytiques et KPI pour la BI';
CREATE SCHEMA IF NOT EXISTS ops     COMMENT 'Qualité des données, logs, quarantaine';

CREATE VOLUME IF NOT EXISTS livraison.landing.raw
  COMMENT 'Fichiers sources bruts déposés par les systèmes amont';
```
